```python
import numpy as np
import matplotlib.pyplot as plt

k = 35000
c = 2900
t_start = 0    
t_end = 20     
m = 300
h = [0.001, 0.005, 0.1]  

def f(t, y):
    y1, y2 = y
    dy1_dt = y2
    dy2_dt =  -(c / m * y2) - (k / m * y1)
    return np.array([dy1_dt, dy2_dt])

def euler_method(h, t_start, t_end):
    t_values = np.arange(t_start, t_end + h, h)
    y_values = np.zeros(len(t_values))
    dy_values = np.zeros(len(t_values))    
    y_values[0] = 0.09
    dy_values[0] = 0

    for i in range(1, len(t_values)):
        y_n = y_values[i - 1]
        dy_n = dy_values[i - 1]        
        y_values[i] = y_n + h * dy_n
        dy_values[i] = dy_n + h * (-(c / m * dy_n) - (k / m * y_n))   
    return t_values, y_values

plt.figure()
for j in h:
    t, y = euler_method(j, t_start, t_end)
    plt.plot(t, y, label=f'Euler\'s Method (h={j})')
plt.xlabel('Time (t)')
plt.ylabel('Displacement (y)')
plt.title('Stability Analysis of Euler\'s Method for a 2nd Order ODE')
plt.legend()
plt.grid(True)

def runge_kutta_4(f, t_start, t_end, h):
    rk_t_values = np.arange(t_start, t_end + h, h)
    rk_y_values = np.zeros((len(rk_t_values), 2))
    rk_y_values[0] = [0.09, 0]
    
    for a in range(1, len(rk_t_values)):
        rk_t = rk_t_values[a - 1]
        rk_y = rk_y_values[a - 1]
        k1 = h * f(rk_t, rk_y)
        k2 = h * f(rk_t + h / 2, rk_y + k1 / 2)
        k3 = h * f(rk_t + h / 2, rk_y + k2 / 2)
        k4 = h * f(rk_t + h, rk_y + k3)
        rk_y_values[a] = rk_y + (k1 + 2 * k2 + 2 * k3 + k4) / 6
    return rk_t_values, rk_y_values [:, 0]
plt.figure()
for step in h:
    r_t, r_y = runge_kutta_4(f, t_start, t_end, step)
    plt.plot(r_t, r_y, label=f'Runge Kutta Method (h={step})')
    
plt.xlabel('Time (t)')
plt.ylabel('Displacement (y)')
plt.title('Stability Analysis of Euler\'s Method for a 2nd Order ODE')
plt.legend()
plt.grid(True)
plt.show()
```
